```python
from astropy.io import ascii
from cxotime import CxoTime
```

```python
ace_table = ascii.read('/data/mta4/Space_Weather/ACE/Data/ace_12h_archive')
```

```python
ace_table
```

# Write test set for ACE P5/P6 spectral index violation alert

```python
#: This rewrites the entire column. The alert only scans the last two hours
ace_table['col13'] = 1e10 #: P5 channel
ace_table['col14'] = 0.5 #: P6 channel
```

```python
filename = "test_p5_p6_spectral_ace_12h_archive"
ascii.write(ace_table, filename, format='no_header', delimiter=' ', overwrite=True)
```

# Write test set for ACE invalid data alert

```python
ace_table['col10'] = -1 #: Change all proton status to -1
```

```python
filename = "test_invalid_ace_12h_archive"
ascii.write(ace_table, filename, format='no_header', delimiter=' ', overwrite=True)
```
